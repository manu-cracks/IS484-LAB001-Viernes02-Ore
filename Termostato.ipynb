```python
def agente_termostato(temperatura_actual, temperatura_objetivo=22, margen=1.0):
    """
    Percepciones (P) : La temperatura medida por el sensor.
    Acciones (A) : Encender/Apagar la calefaccion, Encender/apagar Aire acondicionado, apagado del sistema
    Entorno (E) : Casa, Habitacion, oficina
    Medida de rendimiento (p) : Eficiencia al regular la temperatura, el tiempo que le toma en regular la temperatura, ahorro energetico.
    """
    diferencia = temperatura_actual - temperatura_objetivo
    if diferencia > margen:
        return "Apagar la calefaccion y encender el aire acondicionado"
    elif diferencia < -margen:
        return "Encender la Calefaccion y apagar el aire acondicionado"
    else:
        return "Apagado del sistema"

percepciones = [18.0, 19.5, 21.8, 24.3, 26.0, 23.0, 20.5]
for t in percepciones:
    accion = agente_termostato(t)
    print(f"Percepcion: {t} C -> Accion del agente: {accion}")
```
