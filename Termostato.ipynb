### Agente Inteligente: Termostato (Modelo PEAS)

* **P**ercepciones: La temperatura medida por el sensor.
* **A**cciones: Encender/apagar calefacción, encender/apagar aire acondicionado, apagado del sistema.
* **E**ntorno: Casa, habitación, oficina.
* **R**endimiento (p): Eficiencia al regular la temperatura, tiempo de respuesta y ahorro energético.

### 5.2. Diseño del Agente Termostato (PEAS)

| Percepción (S) | Acciones (A) | Entorno (E) | Objetivo | Medida de desempeño (P) |
|---|---|---|---|---|
| Temperatura ambiente medida por el sensor (°C) | Encender/apagar calefacción, encender/apagar aire acondicionado, apagar sistema | Habitación, oficina o casa | Mantener la temperatura dentro del rango de confort | Tiempo de estabilización y eficiencia en el consumo energético |

### 5.3. Reglas de Decisión

| Condición | Acción a tomar |
|---|---|
| Si la temperatura está muy por encima de la deseada (`diferencia > margen`) | Encender aire acondicionado y apagar calefacción |
| Si la temperatura está muy por debajo de la deseada (`diferencia < -margen`) | Encender calefacción y apagar aire acondicionado |
| Si la temperatura está cerca de la deseada (`dentro del margen`) | Apagado del sistema |

```python
def agente_termostato(temperatura_actual, temperatura_objetivo=22, margen=1.0):
    diferencia = temperatura_actual - temperatura_objetivo
    
    if diferencia > margen:
        return "Apagar la calefaccion y encender el aire acondicionado"
    elif diferencia < -margen:
        return "Encender la Calefaccion y apagar el aire acondicionado"
    else:
        return "Apagado del sistema"

# Prueba del agente con una lista de percepciones
percepciones = [18.0, 19.5, 21.8, 24.3, 26.0, 23.0, 20.5, 21.0, 29.5]

for t in percepciones:
    accion = agente_termostato(t)
    print(f"Percepcion: {t} C -> Accion del agente: {accion}")
```

### 5.6. Reflexión sobre la Racionalidad del Agente

* **¿Es racional?:** Sí, porque bajo sus reglas y con la temperatura medida en el momento, siempre ejecuta la acción adecuada para mantener el confort y optimizar la energía.
* **¿Cuándo dejaría de serlo?:** Si el sensor se descalibra (lee valores falsos) o si el margen de tolerancia es exageradamente amplio (ej. margen = 10 °C), provocando que nunca actúe.
